```python
import requests
import pandas as pd
import sqlite3
import os
from dotenv import load_dotenv

# Load environment variables from the .env file located in the root directory
load_dotenv(dotenv_path="../.env")

API_KEY = os.getenv("CRICBUZZ_API_KEY")
API_HOST = os.getenv("CRICBUZZ_API_HOST", "cricbuzz-cricket.p.rapidapi.com")

headers = {
    "X-RapidAPI-Key": API_KEY,
    "X-RapidAPI-Host": API_HOST
}
```

```python
# Test fetching live matches
url = f"https://{API_HOST}/matches/v1/live"

try:
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    data = response.json()
    
    print("API Connection Successful!")
    # Display the raw JSON structure of the first match type to understand how to parse it
    if "typeMatches" in data and len(data["typeMatches"]) > 0:
        print(data["typeMatches"][0].keys())
    else:
        print("No live matches currently available.")
except Exception as e:
    print(f"API Error: {e}")
```

```python
# Test connecting to the local SQLite database created by our app
db_path = "../data/livestats.db"

try:
    conn = sqlite3.connect(db_path)
    
    # Fetch and display the current players in the database using Pandas
    df = pd.read_sql_query("SELECT * FROM players", conn)
    print("Database Connection Successful!")
    display(df)
    
    conn.close()
except Exception as e:
    print(f"Database Error: {e}")
```
